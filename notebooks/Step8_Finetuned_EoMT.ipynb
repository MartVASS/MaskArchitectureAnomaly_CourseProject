```python
import os
from huggingface_hub import hf_hub_download

FINETUNED_REPO = "daniAimo/eomt_fine_tuned"
FINETUNED_FILENAME = "epoch=4-step=14875.ckpt"
FINETUNED_CHECKPOINT = hf_hub_download(
    repo_id=FINETUNED_REPO,
    filename=FINETUNED_FILENAME,
)

CONFIG = "eomt/configs/dinov2/cityscapes/semantic/eomt_base_640.yaml"
RESULTS_FILE = "results_eomt_finetuned.txt"

print("checkpoint:", FINETUNED_CHECKPOINT)
print("config exists:", os.path.exists(CONFIG))
print("dataset exists:", os.path.exists("data/Validation_Dataset"))
```

```python
open(RESULTS_FILE, "w").close()

!python eval/eval_eomt_finetuned.py \
  --checkpoint "$FINETUNED_CHECKPOINT" \
  --config "$CONFIG" \
  --datasets RoadAnomaly RoadAnomaly21 fs_static LostFound RoadObsticle21 \
  --methods msp entropy maxlogit rba \
  --img-size 640 640 \
  --num-classes 19 \
  --num-q 200 \
  --results-file "$RESULTS_FILE"
```
